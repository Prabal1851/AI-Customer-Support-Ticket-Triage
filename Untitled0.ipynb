```python
import pandas as pd

df_final = pd.read_csv('support_tickets.csv')

print(df_final.shape)
print(df_final['category'].value_counts())
print(df_final['urgency'].value_counts())
df_final.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# Split first, then fit TF-IDF only on training text (avoids data leakage)
X_train, X_test, y_cat_train, y_cat_test, y_urg_train, y_urg_test = train_test_split(
    df_final['text'], df_final['category'], df_final['urgency'],
    test_size=0.2, random_state=42, stratify=df_final['category']
)

vectorizer = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2),      # unigrams + bigrams
    min_df=2,                # ignore terms in fewer than 2 docs
    stop_words='english'
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print("Train shape:", X_train_tfidf.shape)
print("Test shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC

# --- Category classifier ---
category_model = LogisticRegression(max_iter=1000, class_weight='balanced')
category_model.fit(X_train_tfidf, y_cat_train)

# --- Urgency classifier ---
urgency_model = LogisticRegression(max_iter=1000, class_weight='balanced')
urgency_model.fit(X_train_tfidf, y_urg_train)

print("Category model trained.")
print("Urgency model trained.")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# --- Category model evaluation ---
y_cat_pred = category_model.predict(X_test_tfidf)
print("=== CATEGORY CLASSIFICATION REPORT ===")
print(classification_report(y_cat_test, y_cat_pred))

cm_cat = confusion_matrix(y_cat_test, y_cat_pred, labels=category_model.classes_)
plt.figure(figsize=(6,5))
sns.heatmap(cm_cat, annot=True, fmt='d', cmap='Blues',
            xticklabels=category_model.classes_, yticklabels=category_model.classes_)
plt.title('Category Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# --- Urgency model evaluation ---
y_urg_pred = urgency_model.predict(X_test_tfidf)
print("=== URGENCY CLASSIFICATION REPORT ===")
print(classification_report(y_urg_test, y_urg_pred))

cm_urg = confusion_matrix(y_urg_test, y_urg_pred, labels=urgency_model.classes_)
plt.figure(figsize=(6,5))
sns.heatmap(cm_urg, annot=True, fmt='d', cmap='Oranges',
            xticklabels=urgency_model.classes_, yticklabels=urgency_model.classes_)
plt.title('Urgency Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
